```python
# calculator program
a = float(input("enter first number:"))
b = float(input("enter second number:"))
operator = input( "enter operator (+,-,*,/):")
if operator == "+":
    print("answer =", a+b)
elif operator == "-":
    print("answer =", a-b)
elif operator == "*": 
    print("answer =", a*b)
elif operator == "/":
    print("answer =", a/b)

          
```
